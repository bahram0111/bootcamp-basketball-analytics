```python
import requests
import pandas as pd
import re
import time
from bs4 import BeautifulSoup
from tqdm import tqdm
```

```python
url = "https://www.basketball-reference.com/teams/"
response = requests.get(url, headers={"User-Agent": "Mozilla/5.0"}, timeout=20)
print(response.status_code)
soup = BeautifulSoup(response.text, "html.parser")
```

```python
table = soup.find("table", id="teams_active")

data = []

rows = table.find("tbody").find_all("tr", class_="full_table")

for row in rows:

    team_cell = row.find("th")

    if team_cell is None:
        continue

    team_name = team_cell.get_text(strip=True)

    team_link = team_cell.find("a")


    cells = row.find_all("td")

    values = []

    for cell in cells:
        text = cell.get_text(strip=True)
        values.append(text)

    data.append({
        "team_name": team_name,
        "league": values[0],
        "from": values[1],
        "to": values[2],
        "years": values[3],
        "games": values[4],
        "wins": values[5],
        "losses": values[6],
        "win_percentage": values[7],
        "playoffs": values[8],
        "division_championships": values[9],
        "conference_championships": values[10],
        "championships": values[11]
    })

teams_info = pd.DataFrame(data)

teams_info.head()
```

```python
teams_info.to_csv("teams_info.csv", index=False, encoding="utf-8")
```
